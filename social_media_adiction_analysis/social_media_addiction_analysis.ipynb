```python
import pandas as pd

df = pd.read_csv("/content/sample_data/smmh.csv")

df.head()
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt
```

```python
df['7. What social media platforms do you commonly use?'].value_counts().plot(kind='bar')

plt.title("Most Used Social Media Platforms")
plt.xlabel("Platforms")
plt.ylabel("Number of Users")

plt.show()
```

```python
top_platforms = df['7. What social media platforms do you commonly use?'].value_counts().head(10)

top_platforms.plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Most Used Social Media Platforms")
plt.xlabel("Platforms")
plt.ylabel("Number of Users")

plt.xticks(rotation=45)

plt.show()
```

```python
top_platforms = df['7. What social media platforms do you commonly use?'].value_counts().head(5)

top_platforms.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7)
)

plt.title("Top 5 Social Media Platforms")
plt.ylabel("")

plt.show()
```

```python
time_spent = df['8. What is the average time you spend on social media every day?'].value_counts()

time_spent.plot(
    kind='line',
    marker='o',
    figsize=(10,5)
)

plt.title("Average Daily Time Spent on Social Media")
plt.xlabel("Time Range")
plt.ylabel("Number of Users")

plt.xticks(rotation=45)

plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10,5))

sns.countplot(
    x='2. Gender',
    data=df
)

plt.title("Gender Distribution of Social Media Users")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```
